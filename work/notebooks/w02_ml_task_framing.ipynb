## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task type: Scoring**

The goal is to assign each page a content opportunity score based on observed SEO performance, engagement, and trend signals. Pages can then be prioritised for review based on their scores.

This is a scoring problem because the desired output is a continuous measure of how strongly a page appears to be a candidate for content review, rather than simply a yes/no label.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Proxy:** A content opportunity score based on observed SEO signals such as search volume, CTR, average position, and performance trend.

The starter data does not contain a true refresh-success outcome, so this proxy will support **page prioritization** rather than predict whether a refresh will succeed.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Metric: Precision@K**

Measure the proportion of the top-K pages selected by the model that are high-opportunity according to the defined proxy.

This reflects how effectively the scoring system prioritizes pages for limited SEO review time.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule would rely on manually chosen thresholds for individual signals such as position or trend. Content opportunity may instead depend on multiple signals and their interactions.

ML is worth testing because it can learn these patterns from data rather than relying entirely on manually chosen thresholds. However, its value should be demonstrated by comparing it against a strong fixed-rule baseline.
